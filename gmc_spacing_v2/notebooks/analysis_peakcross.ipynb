**Load the Peak-Image Crosses**

```python
cross_dict = {}
indir = '../cloud_cats_fits/homogen_150pc_low_noise/'
first = True
for this_source in source_list:
    
    # Load files
    real_tab = Table.read(indir+this_source+'_realcross.fits')
    mock_tab = Table.read(indir+this_source+'_modelcross.fits')
    
    cross_dict[this_source] = {
        'real':real_tab,
        'model':mock_tab,
    }
    
    #print(this_source, corr_dict[this_source]['nclouds'])
```

```python
png_name = '../plots/cross_onepanel_real_prototype.png'

npanels = 1
ncol = 1
nrow = 1
ticklength = 5

xmin = 0.
xmax = 500
ymin = 0.
ymax = 1.25
xticks = np.arange(xmin,xmax,50)
yticks = np.arange(ymin,ymax,0.1)

mpl_setup('paper-1/1', aspect=3./6., fs=12)

fig, axes = plt.subplots(
    nrow, ncol, sharex='all', sharey='all',
    gridspec_kw={'wspace': 0.1, 'hspace': 0.1})

xlabel = r'Separation [pc]'
ylabel = r'$\left< I_{\rm CO} \right>$ / Peak - real'

fig.text(0.015, 0.5, ylabel, va='center', ha='center', rotation='vertical', fontsize='large')
fig.text(0.5, 0.015, xlabel, va='center', ha='center', rotation='horizontal', fontsize='large')

rows_done = 0
counter = 0

#beam_scale = np.log10(150.)

axes.set_xlim([xmin,xmax])
axes.set_xscale('linear')
axes.set_ylim([ymin,ymax])
axes.set_xticks(xticks)
axes.set_yticks(yticks)
axes.tick_params(direction='inout', length=ticklength)

axes.grid(True, linestyle='dotted',linewidth=0.5,color='black')

plot_main_legend = True
plot_bbox = True
plot_aggregate = True

for this_source in cross_dict.keys():
    this_real_tab = cross_dict[this_source]['real']

    this_x = (this_real_tab['bins_xmid'])
    this_y = (this_real_tab['ymean'])
    this_y = this_y / np.nanmax(this_y)
    
    axes.scatter(this_x,this_y,marker='o',s=20,
                      edgecolor='navy',facecolor='dodgerblue',
                      zorder=2)

    axes.plot(this_x,this_y,zorder=1,linestyle='solid',color='dodgerblue')
        
#if plot_main_legend:
#    axes.legend(fancybox=True, framealpha=0.95, loc=(0.645,0.65), #loc='upper right', 
#        ncol=1, fontsize=10, handlelength=2)    

#fig.tight_layout()
plt.savefig(png_name, bbox_inches='tight')
```
